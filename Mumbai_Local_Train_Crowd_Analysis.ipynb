```python

# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)

print("All libraries imported successfully!")
```

```python
# Step 3: Central Line Dataset

central_data = {
    'Station': [
        'CST', 'Masjid', 'Byculla', 'Dadar', 'Kurla',
        'Ghatkopar', 'Bhandup', 'Mulund', 'Thane',
        'Dombivilli', 'Kalyan', 'Kasara', 'Karjat', 'Khopoli'
    ],

    'Authorised': [
        636661, 245627, 130247, 286394, 348003,
        268225, 175273, 255711, 599249, 283362,
        330942, 13744, 16630, 8058
    ],

    'Unauthorised': [
        0, 0, 2072, 4143, 32927, 0, 0, 0,
        54679, 0, 29406, 3471, 5410, 6256
    ],

    'Total': [
        636661, 245627, 132319, 290537, 380930,
        268225, 175273, 255711, 653928, 283362,
        360348, 17215, 22040, 14314
    ]
}

central_df = pd.DataFrame(central_data)

print("Central Line Passenger Dataset")
display(central_df)
```

```python
# Step 4: Virar-Dahanu Road Line Dataset

dahanu_data = {
    'Station': [
        'Virar', 'Vaitarna', 'Saphale', 'Kelve Road',
        'Palghar', 'Umroli', 'Boisar', 'Vangaon', 'Dahanu Road'
    ],

    'Authorised': [
        318459, 2583, 14035, 9814, 27831,
        2074, 29364, 6702, 36987
    ],

    'Unauthorised': [
        76636, 1107, 0, 0, 0, 321, 1560, 0, 1908
    ],

    'Total': [
        395095, 3690, 14035, 9814, 27831,
        2395, 30924, 6702, 38895
    ]
}

dahanu_df = pd.DataFrame(dahanu_data)

print("Virar-Dahanu Road Line Passenger Dataset")
display(dahanu_df)
```

```python
# Step 5: Harbour Line Dataset

harbour_data = {
    'Station': [
        'Vadala Road', 'GTB Nagar', 'Chembur', 'Govandi',
        'Vashi', 'Nerul', 'CBD Belapur', 'Kharghar',
        'Panvel', 'Airoli', 'Turbe'
    ],

    'Authorised': [
        160645, 97231, 134902, 78349, 219672,
        67948, 169134, 102721, 89536, 68963, 49806
    ],

    'Unauthorised': [
        0, 23871, 38886, 54612, 15097, 35975,
        13717, 9072, 17200, 17192, 15411
    ],

    'Total': [
        160645, 121102, 173788, 132961, 234769,
        103923, 182851, 111793, 106736, 86155, 65217
    ]
}

harbour_df = pd.DataFrame(harbour_data)

print("Harbour Line Passenger Dataset")
display(harbour_df)
```

```python
print("Western Line:", len(df))
print("Central Line:", len(central_df))
print("Virar-Dahanu Line:", len(dahanu_df))
print("Harbour Line:", len(harbour_df))
```

```python
# Step 6: Combine All Railway Lines

df['Line'] = 'Western'
central_df['Line'] = 'Central'
dahanu_df['Line'] = 'Virar-Dahanu'
harbour_df['Line'] = 'Harbour'

# Combine all datasets
final_df = pd.concat(
    [df, central_df, dahanu_df, harbour_df],
    ignore_index=True
)

print("Combined Mumbai Local Train Dataset")
display(final_df)
```

```python
# Step 8: Data Cleaning

# Check missing values
print("Missing Values:")
print(final_df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(final_df.duplicated().sum())

# Check dataset information
print("\nDataset Information:")
final_df.info()
```

```python
# Step 9: Top 10 Busiest Stations

top_stations = final_df.sort_values(
    by='Total',
    ascending=False
)

print("Top 10 Busiest Stations:")
display(top_stations[['Station', 'Line', 'Total']].head(10))
```

```python
# Step 10: Railway Line Comparison

line_comparison = final_df.groupby('Line')['Total'].sum()

print("Total Passengers by Railway Line:")
display(line_comparison.sort_values(ascending=False))
```

```python
# Step 12: Authorised vs Unauthorised Passengers

passenger_comparison = final_df.groupby('Line')[
    ['Authorised', 'Unauthorised']
].sum()

print("Passenger Category Comparison:")
display(passenger_comparison)
```

```python
# Step 13: Average Passengers per Station

average_passengers = final_df.groupby('Line')['Total'].mean()

print("Average Passengers per Station:")
display(average_passengers.sort_values(ascending=False))
```

```python
# Step 14: Top 5 Stations by Unauthorised Passengers

top_unauthorised = final_df.sort_values(
    by='Unauthorised',
    ascending=False
)

print("Top 5 Stations with Highest Unauthorised Passengers:")

display(
    top_unauthorised[
        ['Station', 'Line', 'Unauthorised']
    ].head(5)
)
```

```python
# Step 15: Passenger Distribution Percentage

line_percentage = (
    final_df.groupby('Line')['Total'].sum()
    / final_df['Total'].sum()
) * 100

print("Passenger Distribution by Railway Line (%):")

display(line_percentage.sort_values(ascending=False).round(2))
```

```python
# Step 16: Professional Railway Line Comparison

plt.figure(figsize=(12, 7))

colors = ['#172554', '#2563EB', '#06B6D4', '#F59E0B']

ax = plt.gca()

bars = ax.bar(
    line_comparison.index,
    line_comparison.values,
    color=colors,
    edgecolor='white',
    width=0.6
)

# Add passenger count labels
ax.bar_label(
    bars,
    labels=[f'{v:,.0f}' for v in line_comparison.values],
    padding=6,
    fontsize=10,
    fontweight='bold'
)

plt.title(
    'Mumbai Local Train Passenger Traffic',
    fontsize=19,
    fontweight='bold',
    pad=25
)

plt.suptitle(
    'Railway Line Comparison | Historical 14-Hour Survey',
    fontsize=11,
    color='gray'
)

plt.xlabel('Railway Line', fontsize=12, fontweight='bold')
plt.ylabel('Total Passengers', fontsize=12, fontweight='bold')

ax.set_facecolor('#F8FAFC')
plt.gcf().set_facecolor('white')

plt.grid(axis='y', linestyle='--', alpha=0.25)
ax.set_axisbelow(True)

sns.despine()
plt.tight_layout()
plt.savefig('visualizations/01_passengers_by_line.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Step 17: Top 10 Busiest Stations

top10 = final_df.nlargest(10, 'Total').sort_values('Total')

plt.figure(figsize=(12, 7))
ax = plt.gca()

bars = ax.barh(
    top10['Station'],
    top10['Total'],
    color='#2563EB',
    edgecolor='white',
    height=0.65
)

ax.bar_label(
    bars,
    labels=[f'{v:,.0f}' for v in top10['Total']],
    padding=6,
    fontsize=10,
    fontweight='bold'
)

plt.title(
    'Top 10 Busiest Mumbai Local Railway Stations',
    fontsize=17,
    fontweight='bold',
    pad=20
)

plt.xlabel('Total Passengers (14-Hour Survey)', fontweight='bold')
plt.ylabel('Railway Station', fontweight='bold')

ax.set_facecolor('#F8FAFC')
plt.gcf().set_facecolor('white')

plt.grid(axis='x', linestyle='--', alpha=0.25)
ax.set_axisbelow(True)

sns.despine()
plt.tight_layout()
plt.savefig('visualizations/02_top_10_stations.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Step 18: Passenger Distribution Donut Chart

line_totals = final_df.groupby('Line')['Total'].sum()

colors = ['#172554', '#2563EB', '#06B6D4', '#F59E0B']

fig, ax = plt.subplots(figsize=(10, 8))

wedges, texts, autotexts = ax.pie(
    line_totals,
    labels=line_totals.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    pctdistance=0.78,
    wedgeprops=dict(width=0.42, edgecolor='white', linewidth=3)
)

for text in texts:
    text.set_fontsize(11)
    text.set_fontweight('bold')

for text in autotexts:
    text.set_fontsize(11)
    text.set_fontweight('bold')
    text.set_color('white')

ax.text(
    0, 0,
    f'{line_totals.sum():,.0f}\nTotal Passengers',
    ha='center',
    va='center',
    fontsize=15,
    fontweight='bold',
    color='#172554'
)

plt.title(
    'Passenger Distribution Across Railway Lines',
    fontsize=17,
    fontweight='bold',
    pad=25
)

fig.set_facecolor('white')
plt.tight_layout()
plt.savefig('visualizations/03_passenger_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Step 19: Authorised vs Unauthorised Passengers

passenger_data = final_df.groupby('Line')[
    ['Authorised', 'Unauthorised']
].sum()

fig, ax = plt.subplots(figsize=(11, 7))

colors = ['#2563EB', '#F59E0B']

passenger_data.plot(
    kind='bar',
    stacked=True,
    color=colors,
    edgecolor='white',
    width=0.65,
    ax=ax
)

plt.title(
    'Authorised vs Unauthorised Passenger Distribution',
    fontsize=17,
    fontweight='bold',
    pad=20
)

plt.xlabel('Railway Line', fontsize=12, fontweight='bold')
plt.ylabel('Number of Passengers', fontsize=12, fontweight='bold')

plt.legend(
    title='Passenger Category',
    frameon=False
)

ax.set_facecolor('#F8FAFC')
fig.set_facecolor('white')

plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.25)
ax.set_axisbelow(True)

sns.despine()
plt.tight_layout()
plt.savefig('visualizations/04_authorised_vs_unauthorised.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Step 20: Average Passengers per Station

average_data = final_df.groupby('Line')['Total'].mean()
average_data = average_data.sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(11, 7))

colors = ['#06B6D4', '#2563EB', '#F59E0B', '#172554']

bars = ax.barh(
    average_data.index,
    average_data.values,
    color=colors,
    height=0.6,
    edgecolor='white'
)

ax.bar_label(
    bars,
    labels=[f'{v:,.0f}' for v in average_data.values],
    padding=7,
    fontsize=10,
    fontweight='bold'
)

plt.title(
    'Average Passenger Count per Station',
    fontsize=17,
    fontweight='bold',
    pad=20
)

plt.xlabel('Average Number of Passengers (14-Hour Survey)',
           fontweight='bold')
plt.ylabel('Railway Line', fontweight='bold')

ax.set_facecolor('#F8FAFC')
fig.set_facecolor('white')

plt.grid(axis='x', linestyle='--', alpha=0.25)
ax.set_axisbelow(True)

sns.despine()
plt.tight_layout()
plt.savefig('visualizations/05_average_passengers.png', dpi=300, bbox_inches='tight')
plt.show()
```
